```python
import re
import operator

f = open('Romeo_and_Juliet.txt', 'r', encoding='ascii', errors='replace')

t = f.read().lower()

#wds = re.split(r'\s+', t)
wds = re.findall(r'\b\w+\b', t)


wf = {}
for w in wds:
    if w in wf:
        wf[w] = wf[w] + 1
    else:
        wf[w] = 1

wfs = sorted(wf.items(), key=operator.itemgetter(1), reverse=True)

ml = min(len(wfs), 15)

for i in range(ml):
    print(wfs[i][0] + "\t" + str(wfs[i][1]) + "\t" + f"{(wfs[i][1] / len(wds) * 100):.3}")
```
