# Mini Project 1: Beatles Lyrics Analysis

I plan to analyze the lyrics of The Beatles to see if there are noticeable differences in the words they use throughout their career. I am thinking about comparing their earlier songs to their later songs to see if their word frequencies and writing styles changed over time.

For my data, I plan to use publicly available Beatles lyrics. Some comparisons I am thinking of are comparing word frequencies between early and late Beatles songs, looking for words that are used more frequently during certain periods, and comparing The Beatles to another artist or band to see if there are words that one uses more frequently than the other.

I chose The Beatles because I am interested in their music and they have a large amount of lyrics available to analyze. I think it would be interesting to see if their writing can be distinguished based only on the words they use.
