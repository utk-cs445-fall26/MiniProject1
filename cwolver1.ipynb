# Mini Project 1: Beatles Lyrics Analysis

I plan to analyze the lyrics of The Beatles to see if there are noticeable differences in the words they use throughout their career. I am thinking about comparing their earlier songs to their later songs to see if their word frequencies and writing styles changed over time.

For my data, I plan to use publicly available Beatles lyrics. Some comparisons I am thinking of are comparing word frequencies between early and late Beatles songs, looking for words that are used more frequently during certain periods, and comparing The Beatles to another artist or band to see if there are words that one uses more frequently than the other.

I chose The Beatles because I am interested in their music and they have a large amount of lyrics available to analyze. I think it would be interesting to see if their writing can be distinguished based only on the words they use.


# New Mini Project 1: Book Word Frequency Analysis

I plan to analyze word frequencies in different books to see if there are noticeable differences in the words authors use. I will compare two books by the same author and then compare books written by different authors to see if word frequencies can help distinguish writing styles.

For my same-author comparison, I will use The Time Machine and The War of the Worlds by H. G. Wells. For my different-author comparison, I will compare The Time Machine by H. G. Wells with Frankenstein by Mary Shelley. I will use the publicly available texts from Project Gutenberg as my data.

Some comparisons I plan to make include looking at the most frequently used words in each book and identifying words that appear more frequently in one author's work than another. I will also compare word frequencies as percentages because the books have different lengths.

I chose these books because they provide a good way to compare writing by the same author and by different authors. I think it will be interesting to see how much of the difference in word usage comes from an author's writing style and how much may be influenced by the subject of each book.
