## Goal

The goal of this project is to perform a comparative text analysis of four classic fantasy novels
drawn from two distinct authors: L. Frank Baum (*Oz* series) and Lewis Carroll (*Alice* series).
The analysis aims to uncover stylistic, linguistic, and thematic differences and similarities
between the two authors and across their individual works.

Key questions I plan to investigate:

- **Vocabulary richness**: How large and diverse is each author's vocabulary? Do sequels reuse
  more words than the original?
- **Sentence structure**: Do Baum and Carroll differ in average sentence length and complexity?
- **Word frequency & themes**: What are the most distinctive words for each author/book?
  Do thematic keywords (e.g., *magic*, *dream*, *home*) appear at different rates?
- **Character name frequency**: How often do major characters appear relative to total word count?
- **Sentiment trends**: How does the emotional tone shift across chapters within each book,
  and does one author skew more positive/negative than the other?

---

## Data Sources

I will use four public-domain fantasy books from Project Gutenberg:

- L. Frank Baum, *The Wonderful Wizard of Oz*: https://www.gutenberg.org/ebooks/55
- L. Frank Baum, *The Marvelous Land of Oz*: https://www.gutenberg.org/ebooks/54
- Lewis Carroll, *Alice's Adventures in Wonderland*: https://www.gutenberg.org/ebooks/11
- Lewis Carroll, *Through the Looking-Glass*: https://www.gutenberg.org/ebooks/12

All texts are freely available as plain-text `.txt` files via the Project Gutenberg website.
I will download them programmatically using the `requests` library and strip the standard
Gutenberg header/footer boilerplate before analysis.

---

## Approach

1. **Data retrieval**: Download plain-text versions of each book from Project Gutenberg.
2. **Preprocessing**: Strip Gutenberg headers/footers, tokenize text, normalize case,
   and remove punctuation/stop words where appropriate.
3. **Exploratory analysis**: Compute word counts, unique vocabulary size, type-token ratio,
   and average sentence length for each book.
4. **Frequency analysis**: Build word frequency distributions and compare top-N words
   within and across authors using bar charts and word clouds.
5. **TF-IDF**: Use TF-IDF to identify the most distinctive terms per book.
6. **Sentiment analysis**: Apply a lexicon-based sentiment scorer (e.g., VADER or TextBlob)
   chapter-by-chapter to plot emotional arcs.
7. **Comparison & visualization**: Summarize findings with plots (matplotlib/seaborn)
   and interpret what the numbers reveal about each author's style.
