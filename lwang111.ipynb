# Assignment details

## Goals

Can the frequency of the most common words help reveal the subject matter and themes of a text, and offer hints about the author's purpose?

Can I make a general dectuction on the author$'s purpose by the top 10 most frequent words of their work? How close is my deduction to a summary of the book?

To do this, we need to filter out words that are common across nearly all English text and therefore don't distinguish one text's content from another's — words like 'and', 'the', 'or', etc. (These "stop words" still serve a grammatical function; they're just not distinctive.)

Filtering them out will help surface words that are specific to a given text, which can point to its subject matter and, more speculatively, the author's intent (to inform, persuade, entertain, etc.).

The two works by a single author I chose are [Great Expectations](https://www.gutenberg.org/ebooks/1400) and [A Tale of Two Cities](https://www.gutenberg.org/ebooks/98).

The works of two different authors I will compare are [Pride and Prejudice](https://www.gutenberg.org/ebooks/1342) and [Great Expectations](https://www.gutenberg.org/ebooks/1400)

I will then compare which words Charles Dicken's uses that Jane Austen might not use as much from specific works.

## Steps to achieve goal

- Fetch Book contents from Project Gutenburg
- Filter out sections that are not the book's conent (Inroduction, Table of Contents, .etc)
- Filter out "stop words" from the sections to get distinctive words
- Count frequency of words for each book
