## Goals

The goal of this assignment is to compare word usage across different texts and authors.

I will:
- Compare word frequencies between two works written by the same author.
- Compare word frequencies between works written by two different authors.
- Look for words that one author uses more frequently than another.

To do this, I will use publicly available text data, such as books from Project Gutenberg. I will process the texts by tokenizing the words, removing punctuation and common stop words, and calculating word frequencies. I will then compare the results using word counts and relative frequencies.
