# Mini Project 1 Proposal

## Goal

For this mini project, I want to explore how different cybersecurity organizations describe the same vulnerabilities. My main research question is:

**How do word-use patterns differ between CISA and NVD descriptions of the same cybersecurity vulnerabilities?**

The goal is to treat the vulnerability descriptions as operational text data and look for measurable patterns in the words used by each source. This follows the project objective of comparing patterns in how authors or organizations use words rather than only focusing on the meaning of the text.

## Data Sources

I plan to use two publicly available cybersecurity data sources:

1. **CISA Known Exploited Vulnerabilities (KEV) Catalog**
2. **National Vulnerability Database (NVD)**

Both sources contain vulnerability information associated with CVE identifiers.

I plan to match vulnerabilities using their **CVE IDs** so that I can compare CISA and NVD descriptions of the same vulnerability. This should make the comparison more meaningful because both descriptions will be discussing the same underlying security issue.

## Approach

I plan to collect a sample of CVEs that appear in both CISA and NVD and retrieve the vulnerability descriptions from each source.

Using Python, I will clean and analyze the text. The analysis may include:

- Converting text to lowercase
- Removing punctuation and unnecessary characters
- Removing common stop words
- Calculating word frequencies
- Calculating normalized word frequencies to account for different text lengths
- Comparing the most frequently used words between CISA and NVD
- Comparing description lengths and vocabulary usage
- Identifying words that appear more frequently in one source than the other
- Visualizing some of the differences using graphs or charts

## Expected Comparison

The main comparison will be between the word-use patterns in CISA descriptions and NVD descriptions for matched CVEs.

For example, I want to investigate whether one source tends to use certain cybersecurity, technical, or action-oriented words more frequently than the other.

The purpose is not to determine which organization provides better descriptions, but rather to observe and measure differences that appear in the text.

## Limitations

The results will only show patterns in the descriptions included in my dataset. Differences in word frequencies may reflect different writing styles, purposes, or documentation practices, but the text alone may not be enough to determine why those differences exist.

Because of this, I will separate what can be directly observed from the data from conclusions that the evidence cannot support.
