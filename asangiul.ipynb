Goals and Ways to Achieve those Goals comparing two works of literature: 

My goal for this project is to compare 'Sense and Sensibility' and 'Pride and Prejudice' by Jane Austen. I would like to see how much her debut album 'Sense and Sensibility' language from 1811 compares to her later 'Price and Prejudice' in 1813. I would like to know if her writing style remained the same or if she changed between the two books. For example I'd like to find out if she got more comforable in her writing to use bolder words or different syntax. This can be achieved by: highlighting which words remained common between the two, anaylzing the difference between the most common versus the least common language, and also checking the outliers to see if they are similar or completely different. 

I would also like to compare Susan Ferrier's book from 1818, "Marriage" to one of Jane Austen's works. Ferrier is known as the "Scottish Austen" and I would like to see how different/similar their writing styles are for the same time frame. It will also be interesting to see how different her Scottish regional identity differs from Austen's English high fashion.
