For this project I reviewed the USDA AMS Market News Terminal Market Prices of Fruits and Vegetables. This dataset contains daily wholesale prices of fruits and vegetables sold at major US terminal markets (like New York, L.A., Boston, etc.). These are categorized as: commodity, variety, origin, package type, and quality grade. An interesting goal could be to check how wholesale prices for the same item differ across terminal markets in different regions on the same date, and what that reveals about regions supply chains. A way to achieve this goal is to choose a date, then check each region and find their common fruits and vegetables. From there we can compare the differences and analyze the data.
