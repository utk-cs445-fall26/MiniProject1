First, I am going to compare two books by Lewis Carroll to count their word frequencies to see if there is any similarities between words preferred. Then, I will compare one of the first two to Dracula by Bram Stoker to see if there is any difference between the two. Hopefully, this will reveal if a certain author has a preference for certain words over another or if they are the similar.

This will be acheived by first downloading a plain text of each work. Then, the word frequencies will be calculated on each work. Two from the same author will be compared, then one from the first two and one from a new author.

```python
import requests, re, nltk
from bs4 import BeautifulSoup
from nltk import clean_html
from collections import Counter
import operator

# urls of texts to compare
wonderland = 'https://www.gutenberg.org/cache/epub/11/pg11.txt'
looking_glass = 'https://www.gutenberg.org/cache/epub/12/pg12.txt'
dracula = 'https://www.gutenberg.org/cache/epub/345/pg345.txt'


# words to filter out, both a default set in the english language, and custom set
# need to download "stopwords" to work
stop_words = nltk.corpus.stopwords.words('english') + [
 'ut', '\'re','.', ',', '--', '\'s', '?', ')', '(', ':', '\'',
 '\"', '-', '}', '{', '&', '|', u'\u2014', '“', '*', '$', '[', ']', '@', '>', '<', ';', '=', '/', '', ' ']

# function to remove html mark up
def cleanHtml(html):
    soup = BeautifulSoup(html, 'html.parser')
    return soup.get_text()

# function to remove special characters, quotes, etc. from each word
def cleanWord(w):
    # remove any character between the brackets []
    wn = re.sub('[,".\'&|:@><*;“/=]', "", w)
    # remove numbers
    return re.sub('^[0-9.]*$', "", wn)

# function to that returns a list of words with their frequencies and the total words counted
def getWordFeq(url, cap):
    r = requests.get(url)

    # clean the text of html, split into words based on white space, and clean
    text = cleanHtml(r.text).lower()

    words = re.split('\\s+', text)
    for i in range(len(words)):
        words[i] = cleanWord (words[i])

    # count all the words, remove stop words, and total up the amount of words
    word_feq = Counter(words)

    for i in stop_words:
        word_feq.pop(i, None)
    
    total_words = 0
    for i in word_feq:
       total_words += word_feq[i]

    # sort the list, cap it to a set number
    word_feq_sorted = sorted(word_feq.items(), key = operator.itemgetter(1), reverse=True)
    max_length = min(len(word_feq_sorted), cap)

    return (word_feq_sorted[0:max_length][::-1], total_words)

    # [::-1]

max = 15

(word_freq_wonderland, total_words_wonderland) = getWordFeq(wonderland, max)
(word_freq_looking_glass, total_words_looking_glass) = getWordFeq(looking_glass, max)
(word_freq_dracula, total_words_dracula) = getWordFeq(dracula, max)

```

```python
import numpy as np
import pylab
import matplotlib.pyplot as plt

def plotList(list, title):
    words, count = zip(*list)
    plt.barh(words, count)
    plt.title(title)
    plt.xlabel('Count')
    plt.ylabel('Word')
    plt.show()

def plotTwoLists(list1, list2, title1, title2):
    words1, count1 = zip(*list1)
    words2, count2 = zip(*list2)
    fig, plots = plt.subplots(1, 2, figsize=(10, 5))

    plots[0].barh(words1, count1, color='green')
    plots[0].set_title(title1)

    plots[1].barh(words2, count2, color='pink')
    plots[1].set_title(title2)

    plt.tight_layout()
    plt.show()  

#plotList(word_freq_wonderland, "Wonderland")
#plotList(word_freq_looking_glass, "Looking Glass")
#plotList(word_freq_dracula, "Dracula")

print("Comparison of word frequencies between Alice's Adventures in Wonderland and Through the Looking-Glass by Lewis Carroll:")
plotTwoLists(word_freq_wonderland, word_freq_looking_glass, "Alice's Adventures in Wonderland", "Through the Looking-Glass")
print("Comparison of word frequencies between Alice's Adventures in Wonderland by Lewis Carroll and Dracula by Bram Stoker:")
plotTwoLists(word_freq_wonderland, word_freq_dracula, "Alice's Adventures in Wonderland", "Dracula")
```
