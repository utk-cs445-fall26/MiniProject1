# MiniProject 1 Proposal

I want to study electricity demand using data from the U.S. Energy Information Administration (EIA). 
The dataset contains hourly electricity demand, demand forecasts, generation, and interchange data for different balancing authorities in the United States.

My main question is: 
which balancing authority has the smallest gap between forecasted demand and actual demand over the last five years?

I plan to use the EIA API or bulk download to collect demand and day-ahead forecast data. I will compare forecasted demand with actual demand for different balancing authorities and calculate the forecast error. I may also use plots to compare how accurate different regions are over time.


One possible problem is that some records may be missing or reported incorrectly. The dataset also does not explain why electricity demand changes or what type of customers are using the electricity.
