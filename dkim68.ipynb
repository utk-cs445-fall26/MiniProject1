# Assignment 1 Proposal

## Data
I will use three books from Project Gutenberg:

- *Pride and Prejudice* by Jane Austen
- *Emma* by Jane Austen
- *Adventures of Huckleberry Finn* by Mark Twain

## Goal
First, I will compare word frequencies between two books written by Jane Austen. Then I will compare Austen's word usage with Mark Twain's word usage.

I also want to find words that Austen uses more frequently than Twain.

## How to Achieve the Goal
1. Download and clean the text of each book.
2. Remove HTML, punctuation, and common stop words.
3. Count the frequency of each remaining word.
4. Compare *Pride and Prejudice* and *Emma*.
5. Compare Austen's books with *Adventures of Huckleberry Finn*.
6. Normalize word frequencies so books of different lengths can be compared fairly.
7. Find words that appear relatively more often in Austen's writing than in Twain's writing.
