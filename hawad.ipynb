# Mini Project 1: Movie Review Sentiment Analysis

## Question 
How does text review of movies relate to how they are ranked positive or negative

## Goal 
The goal is to analyze movie text review and identify the diffrences between the positive and negative reviews.

## Data source
I will use the large movie review dataset from stanford with 50,000 labeled reviews.


## Approch
I will clean and process the dataset and analyze/compare the words used in positive and negative reviews.


## Compareisons:
I plan to compare:
Most common words in positive and negative reviews
Average review length
Sentiment scores between positive and negative reviews
I will use tables to present the results and discuss what the differences in language suggest about positive and negative movie reviews.
