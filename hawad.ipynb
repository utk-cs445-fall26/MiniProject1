# Mini Project 1 — Authorship Analysis

## Research Question

Can word frequency patterns help distinguish between Jane Austen and Charles Dickens?

## Goal

I want to compare the words used by Jane Austen and Charles Dickens to see whether they have different patterns of word usage.

## Data Sources

I will use four books from Project Gutenberg:

* Jane Austen — *Pride and Prejudice*
* Jane Austen — *Sense and Sensibility*
* Charles Dickens — *A Tale of Two Cities*
* Charles Dickens — *Great Expectations*

## Approach

I will use Python to clean the text and count how often each word appears. I will remove common words, punctuation, and other unnecessary text.

I will then compare the most frequent words in the books and look for words that are used more often by one author than the other.

## Comparisons

I will compare:

1. The word frequencies between books by the same author.
2. The word frequencies between Jane Austen and Charles Dickens.
3. Words that are used more frequently by one author than the other.

I will use tables and graphs to show the results.
