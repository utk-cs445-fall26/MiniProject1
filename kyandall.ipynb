# Mini Project 1 - Cybersecurity vs. Technology News

## Goal
Compare the language used in cybersecurity news articles with the language
used in general technology news articles.

## Data
I will collect publicly available articles from cybersecurity and general
technology news sources and separate them into two datasets.

## Analysis
I will compare characteristics such as common words, word frequencies,
article length, and potentially sentiment.

## Comparison
The main comparison will be cybersecurity articles versus general technology
articles to determine whether the two categories have noticeable differences
in their language.
