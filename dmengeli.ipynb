# Assignment Goals

1. Compare word frequencies between two works of a single author.
2. Compare word frequencies between works of two authors.
3. Determine whether there are words preferred by one author but used less frequently by another author.

# Proposed Authors and Works

For this assignment, I think a good author to choose would be **William Shakespeare**. He has a wide vocabulary and can be somewhat inconsistent with his more flexible grammar, which would make it interesting to see which patterns stick across his works.

I would suggest using two of his most famous works:

- *Romeo and Juliet*
- *Hamlet*

# Comparing Two Authors

For comparing word frequencies between works of two different authors, I think it would be interesting to compare **William Shakespeare** and **Edgar Allan Poe**.

Poe's works were written roughly 200 years after Shakespeare's, which would make it interesting to examine how writing style and grammatical conventions changed over time.

Works by both authors are publicly accessible and can be found online.
