Goal: Compare word frequencies in written texts to see whether word usage can distinguish one author from another.

Questions: (1) How do word frequencies differ between two works of the same author? (2) How do they differ between two different authors? (3) Are there words strongly preferred by one author but rarely used by the other? 

Data source: Project Gutenberg plain-text books --- e.g., Jane Austen (Pride and Prejudice, Emma) and Mark Twain (Adventures of Huckleberry Finn, Tom Sawyer), retrieved via HTTP with Python requests.

Comparisons: Within-author (Austen vs. Austen) vs. between-author (Austen vs. Twain) top-word frequency distributions, after removing stop words. Extra: test whether the frequency of a specific word (e.g., "would") follows a binomial distribution and differs significantly between books/authors.
