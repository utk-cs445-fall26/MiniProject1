Goal: Analyze government takedown notices sent to GitHub (github/gov-takedowns) to understand how internet censorship requests differ across countries and change over time.

Questions: (1) Which countries send the most takedown requests, and how has the volume changed by year?

Data source: The github/gov-takedowns repository, which contains verbatim takedown notices organized by country and year. I will clone the repo, extract country/year metadata from file paths, parse the notice text (markdown and PDF), and semantic analysis.

Comparisons: Across countries (Russia vs. China vs. others) and across years.
