# Mini Project 1: Air Quality Sensor Performance and Urban Pollution Patterns



## Project Goal



This project will analyze the **Air Quality dataset** from the UCI Machine Learning Repository. The data were collected by an air-quality monitoring system placed at road level in a polluted area of an Italian city. The system included five metal-oxide chemical sensors and a nearby certified analyzer that measured pollutants such as carbon monoxide (CO), benzene, nitrogen oxides (NOx), and nitrogen dioxide (NO2). Temperature and relative humidity were also recorded. Each row represents one hour of averaged measurements.



My main goal is to investigate whether the experimental sensor responses remained consistent with the certified reference measurements throughout the monitoring period. This may reveal evidence of sensor drift or changes in sensor performance over time. I will also examine how pollutant concentrations varied by time of day during the year of monitoring.



## Research Questions



1. How did the relationship between the metal-oxide sensor responses and the certified reference pollutant measurements change over time?

2. Did the differences between sensor responses and reference measurements become larger or smaller during the monitoring period?

3. How did measured air-pollutant concentrations vary by hour of day, day of week, or season?

4. How were temperature and relative humidity associated with sensor responses and pollutant measurements?



## Data Sources



- [UCI Machine Learning Repository: Air Quality](https://archive.ics.uci.edu/dataset/360/air+quality)

- [Kaggle: Air Quality Data Set](https://www.kaggle.com/datasets/fedesoriano/air-quality-data-set)



The dataset is available as CSV and XLSX files. It originated from a field deployment used by Saverio De Vito and collaborators to study the calibration of low-cost pollution sensors. The measurements represent real urban conditions, although the deployment itself was associated with a research project rather than a purely operational monitoring program.



## Planned Approach



- Download and load the air-quality data into a pandas DataFrame.

- Combine the date and time fields into a single datetime value and sort the observations chronologically.

- Replace the dataset's `-200` missing-value codes with proper missing values.

- Check data types, missingness, date coverage, and unusual observations.

- Compare each experimental sensor response with its related certified reference measurement using correlations, scatterplots, and time-based summaries.

- Divide the monitoring period into monthly or seasonal intervals to determine whether these relationships changed over time.

- Group reference pollutant measurements by hour of day and day of week to identify recurring urban pollution patterns.

- Examine the relationships among pollutant levels, temperature, and relative humidity.

- Present the results with time-series plots, grouped summaries, correlation heatmaps, and other appropriate visualizations.



## Expected Challenges and Limitations



The exact monitoring location is not provided beyond being at road level in a polluted area of an Italian city. Traffic volume, construction, road closures, holidays, fires, weather, and other unrecorded events may have affected the measurements. Missing observations are encoded as `-200` and must be handled before analysis. In addition, each metal-oxide sensor was nominally targeted toward a pollutant but may also respond to other gases. Changes in a sensor's relationship with a reference measurement could suggest sensor drift, but they could also reflect changing atmospheric conditions, temperature, humidity, or cross-sensitivity. Therefore, the analysis can identify patterns and possible drift, but it cannot establish a single cause without additional calibration and site information.



## Intended Outcome



The analysis will show whether low-cost sensor responses tracked certified pollutant measurements consistently and when urban pollutant levels were typically highest. These findings may help explain both the usefulness and limitations of experimental air-quality sensors in real-world monitoring.
