# Part 2: Computing Average Word Lengths

`MiniProject1.ipynb`'s utility functions are redeclared in this notebook in the cell below. Then, the average word length is computed for each book, and its value is added to its corresponding genre's list.

```python
import requests, re, nltk
from bs4 import BeautifulSoup
from nltk import clean_html
from collections import Counter
import operator

nltk.download('stopwords')

# utility functions and variables from MiniProject1.ipynb
stop_words = nltk.corpus.stopwords.words('english') + [
 'ut', '\'re','.', ',', '--', '\'s', '?', ')', '(', ':', '\'',
 '\"', '-', '}', '{', '&', '|', u'\u2014' ]

def cleanHtml (html):
    from bs4 import BeautifulSoup
    soup = BeautifulSoup(html, 'html.parser')
    return soup .get_text()

def cleanWord (w):
    wn = re.sub('[,"\\.\'&\\|:@>*;/=]', "", w)
    return re.sub('^[0-9\\.]*$', "", wn)
       
def get_wf (URL):
    r = requests .get(URL)
    t = cleanHtml (r .text) .lower()
    wds = re .split('\\s+',t)
    for i in range(len(wds)):
        wds [i] = cleanWord (wds [i])
    wf = Counter (wds)
    for k in stop_words:
        wf. pop(k, None)
    tw = 0
    for w in wf:
       tw += wf[w]
    wfs = [x for x in sorted(wf .items(), key = operator.itemgetter(1)) if x[0] != ''] # modified to remove empty string
    return (wfs[:][::-1], tw)

(wf_ee, tw_ee) = get_wf('https://www.gutenberg.org/cache/epub/14838/pg14838-images.html')

print(wf_ee[:5])
print(tw_ee)
```

```python
import re
import math
import os
from time import sleep

genres_awl = {}

# iterate through of the books
ctr = 0
for genre in METADATA:
    genres_awl[genre] = []
    for title, authors, url in METADATA[genre]:
        print(f"({math.ceil(ctr * 100 / num_books)}%) Processing \"{title}\" by {authors}...".ljust(150, " "), end='\r')

        # start computing and storing statistics
        word_freqs, total_word_count = get_wf(url)
        weighted_sum = 0
        for word, word_count in word_freqs:
            weighted_sum += len(word) * word_count / total_word_count
        genres_awl[genre].append(weighted_sum)

        ctr += 1

print("(100%) Finished processing books!".ljust(100, " "))
```

## Part 3: Graphing Results

Genres are plotted against their average word lengths for a visibly intuitive result.

```python
import numpy as np
import matplotlib.pyplot as plt

genres = [genre for genre in METADATA]
values = [np.mean(genres_awl[genre]) for genre in genres]
errors = [np.std(genres_awl[genre]) for genre in genres]

# Create bar chart with error bars
plt.figure(figsize=(8, 6))
plt.bar(genres, values, yerr=errors, capsize=8)

# Add labels and title
plt.title('Average Word Length per Book Genre', fontsize=14)
plt.xlabel('Genre')
plt.ylabel('Average Word Length')

# Show plot
plt.show()
```

## Conclusion

While the mystery, adventure, and fantasy genres all had very similar average word lengths, the children's genre's was significantly lower. Unsurprisingly, these results suggest that children's books often have a simpler vocabulary and often use shorter words.
